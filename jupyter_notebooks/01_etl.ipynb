The 'zero' Cholesterol rows contained 152 patients with HeartDisease (88.4% of the zero-cholesterol group) and 20 patients (11.6%) with no HeartDisease.  This compares with all 918 patients who were split with 508 patients having heart disease (55.3% of the total) and 410 (44.7%) without heart disease.  Thus, patients in the subgroup with a cholesterol value of zero appear to be more likely to have heart disease than patients overall. To remove these patients from the dataset could bias the sample, removing a large number of patients that one would want to study i.e. those patients with heart disease.

```python
(df["RestingBP"] == 0).sum()
```

```python
df_clean = df.copy()
df_clean["Cholesterol_Missing"] = (df_clean["Cholesterol"] == 0).astype(int)
df_clean["Cholesterol_Missing"].sum()
```

```python
df_clean.columns
```

```python
cholesterol_median = df_clean.loc[df_clean["Cholesterol"] != 0, "Cholesterol"].median()
print(cholesterol_median)
```

```python
df_clean["Cholesterol"] = df_clean["Cholesterol"].replace(0, cholesterol_median)
```

```python
# Has replacing cholesterol zeroes with median worked as expected?
(df_clean["Cholesterol"] == 0).sum()    #   0
df_clean["Cholesterol_Missing"].sum()   # 172
(df_clean["Cholesterol"] == 237).sum()  # 178
df_clean["Cholesterol"].median()        # 237.0
# Checks after replacing zeroes with median all gave expected results, and revealed that there were six original values of 237.0 prior to the replacement.
```

## The Fate of RestingBP

The search for zeros revealed that the vast majority of them occurred within 172 rows, in the Cholesterol field.  However, there was also a single instance of a zero located in the RestingBP field, which had to be erroneous as a resting blood pressure of zero is incompatible with life.  This posed a problem of how it was to be tackled.  The isolated row could either be discarded, with potential loss of useful information, or it could be retained and a value for the resting blood pressure imputed using the median of the non-zero values in the field.

I decided to keep the row and impute a value for resting blood pressure, despite the corresponding value for Cholesterol also having been missing,  with a zero value, originally.  I believe imputation for RestingBP, with the associated retention of other data values in the row, outweighs the argument for removing a partial record with two  erroneous data values.

Following replacement by the median value, 119 instances of the median were found in RestingBP, and the median was 130.  The shape of the dataset at this stage was (918, 13) i.e. all original rows are retained, and an additional 'flag' column, Cholesterol_Missing, has been created.

```python
df_clean[df_clean["RestingBP"] == 0]
```

```python
resting_BP_median = df_clean.loc[df_clean["RestingBP"] != 0, "RestingBP"].median()
print(resting_BP_median)
df_clean["RestingBP"] = df_clean["RestingBP"].replace(0, resting_BP_median)
# Has replacing resting BP zero with median worked as expected?
#(df_clean["RestingBP"] == 0).sum()   #   0
#(df_clean["RestingBP"] == resting_BP_median).sum()  # 119
#df_clean["RestingBP"].median()        # 130.0
df_clean.shape
# Checks after replacing zeroes with median all gave expected results, and revealed that there were six original values of 237.0 prior to the replacement.
```

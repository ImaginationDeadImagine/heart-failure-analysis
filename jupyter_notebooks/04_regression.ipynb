# Section 1

### Load the cleaned data

```python
import pandas as pd
df = pd.read_csv("data/processed/heart_clean.csv")
df.shape
```

```python
X = df.drop(columns="MaxHR")
y = df["MaxHR"]
```

```python
X.shape
```

```python
y.shape
```

```python
"MaxHR" in X.columns
```

```python
"HeartDisease" in X.columns
```

The features, X, are from the original DataFrame minus the MaxHR column. The target, y, contains the original MaxHR column and nothing else.  This separation means original data can be processed using the features without the target values being used as input to the process.  The learning stage of the regression process does, however, have access to the target as part of building the model.

I decided to keep HeartDisease in the model as a feature because, as the t-test showed, it is related to MaxHR.  It also keeps the model simpler.  A limitation of this approach is that a diagnosis may not be known at the point of prediction.  Also, other features (ExerciseAngina, Oldpeak, ST_Slope) are only recorded during the same exercise test as MaxHR.  Whilst a purist might remove them, that would remove several informative columns, so I have decided to retain them as features.

### One-Hot Encoding

```python
X_encoded = pd.get_dummies(X, drop_first=True, dtype=int)
X_encoded.shape
```

```python
X_encoded.head()
```

```python
"MaxHR" in X_encoded.columns
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

The data was split according to the pattern set out for classification, except that the split was not stratified.  Unlike HeartDisease, there are no obvious groups to balance here; MaxHR is a number, not a category.  There are enough cases here, 918, that a random split should result in a similar spread of values between the train and test groups.

### Definition of Statistical Terms

| Term | Meaning |
| :--- | :------ |
| Linear regression | A model that adds up points from each feature to give a number, such as a predicted heart rate.  The total is a number, unlike the yes/no result of classification. |

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
y_pred.shape
```

```python
y_pred[:5]
```

```python
from sklearn.metrics import r2_score, mean_absolute_error

r2 = r2_score(y_test, y_pred)
r2
```

```python
mae = mean_absolute_error(y_test, y_pred)
mae
```

### Definition of Statistical Terms

| Term | Meaning |
| :--- | :------ |
| Mean absolute error (MAE) | The average size of the model's misses, in the target's own units. |
| $R^2$ ("R-squared") | How much better the model is than always guessing the average, where 0 means no better and 1 means perfect. |
| $R^2$ between 0 and 1 | The share of the natural variation in MaxHR that the model explains. |
| $R^2$ below 0 | Worse than the lazy guess of always guessing the average. |

### Calculating a Baseline

```python
guess_list = [y_train.mean()] * len(y_test)
```

```python
baseline_mae = mean_absolute_error(y_test, guess_list)
baseline_mae
```

For prediction of MaxHR based on the model features, the $R^2$ score was 0.308.  The model explains roughly a third of the variation in maximum heart rate.  The model has positive predictive power but it is only a modest result.

The model's predictions of MaxHR were out by 16.6 beats per minute on average.  A baseline model that always predicts the average was out by 21.4 beats per minute.  The model's average error, then, is about 22% smaller.  This compares favourably to the baseline but only modestly.

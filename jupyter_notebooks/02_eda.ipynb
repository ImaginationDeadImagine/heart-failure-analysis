```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/processed/heart_clean.csv")
df.shape
```

### Definition of statistical terms

| Term | Meaning |
| :--- | :------ |
| Distribution | The pattern of values such as how spread out they are, and where they sit. |
| Skew | A measure of assymetry in the distribution.  The direction is named for the tail that stretches out further. |
| Left-skewed |  The longer tail is to the left. |
| Mode | The most frequent value in the histogram i.e. the value with the tallest bar. |

### Age distribution

```python
sns.histplot(data=df, x="Age")
plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Number of patients")
```

The histogram of the age of the patients shows a left-skewed distribution with a peak, or mode, of about 55 years of age.  The age range of the patients who were assessed for heart disease is approximately from 30 to 75 years, and findings will therefore apply to this age group.

### Heart disease

```python
sns.countplot(data=df, x="HeartDisease")
plt.title("Heart disease vs. no heart disease")
plt.xlabel("Heart disease category (0 = no heart disease, 1 = heart disease)")
plt.ylabel("Number of patients")
```

The bar chart above shows that the dataset contained more patients with heart disease (category 1) than without (category 0).  Despite that, the two groups are fairly balanced, containing approximately 400 and 500 patients in the 'no heart disease' and 'heart disease' categories respectively.  This represents about 44% and 56% of the overall total, so the groups can be regarded as fairly balanced.

### Definition of Statistical Terms

| Term | Meaning |
| :--- | :------ |
| Quartile | A boundary on a box plot that shows where 25%, 50% or 75% of values fall beneath.<br>The 25% boundary is the lower border of the box, the 50% boundary is the central line<br>within the box, and the 75% boundary is the upper border of the box.    |
| Interquartile range (IQR) | The interquartile range refers to the 50% of values that fall within the upper and<br>lower borders of the box.    |
| Whisker | A whisker is a line drawn parallel to the upper or lower border of the box, typically at a<br>distance of 1.5 times the IQR from the border.     |
| Outlier | Outliers are typically defined as values falling outside the upper or lower whiskers.     |


### Relationship between heart disease category and maximum heart rate

```python
sns.boxplot(data=df, x="HeartDisease", y="MaxHR")
plt.title("Comparison of Heart Disease Categories by Maximum Heart Rate")
plt.xlabel("Heart disease category (0 = no heart disease, 1 = heart disease)")
plt.ylabel("Maximum heart rate (beats per minute)")
```

The plot of maximum heart rate achieved by the group with heard disease versus the group without, shows that the group without heart disease had a higher median for maximum heart rate.  There is some overlap between the middle 50% of values (the interquartile range) between the two groups but it is not great.

Three outliers in category 0 (no heart disease) are found and are below the lower whisker, whilst in category 1 (heart disease) two outliers lie below the lower whisker and one is above the upper whisker.

The plot suggests that maximum heart rate achieved is greater in those without heart disease.
